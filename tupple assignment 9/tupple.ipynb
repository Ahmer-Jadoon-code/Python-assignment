```python
# Q8
employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)

temp_list = [
    (employees[0][1], employees[0][0]),
    (employees[1][1], employees[1][0]),
    (employees[2][1], employees[2][0])
]
temp_list.sort(reverse=True)

sorted_employees = (
    (temp_list[0][1], temp_list[0][0]),
    (temp_list[1][1], temp_list[1][0]),
    (temp_list[2][1], temp_list[2][0])
)
print(sorted_employees)
```

```python
# Q9
matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))
diagonal = (matrix[0][0], matrix[1][1], matrix[2][2])
diagonal_product = diagonal[0] * diagonal[1] * diagonal[2]
row_sums = (sum(matrix[0]), sum(matrix[1]), sum(matrix[2]))

print("Diagonal:", diagonal)
print("Product:", diagonal_product)
print("Row Sums:", row_sums)
```

```python
# Q10
records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)
total_inventory_value = (records[0][1] * records[0][2]) + (records[1][1] * records[1][2]) + (records[2][1] * records[2][2])

price1 = (records[0][1], records[0][0])
price2 = (records[1][1], records[1][0])
price3 = (records[2][1], records[2][0])
most_expensive = max(price1, price2, price3)[1]

reversed_names = (records[2][0], records[1][0], records[0][0])

print("Value:", total_inventory_value)
print("Most Expensive:", most_expensive)
print("Names Reversed:", reversed_names)
```
