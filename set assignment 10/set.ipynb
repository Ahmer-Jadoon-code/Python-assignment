```python
# Q2
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}
union_set = a | b
intersection_set = a & b
difference_set = a - b
symmetric_diff = a ^ b
print(union_set, intersection_set, difference_set, symmetric_diff)
```

```python
# Q4
python_students = {"Ali", "Sara", "Ahmed", "Zain"}
sql_students = {"Sara", "Ahmed", "Hamza", "Usman"}
both = python_students & sql_students
exactly_one = python_students ^ sql_students
print(both, exactly_one)
```

```python
# Q5
required_skills = {"Python", "SQL"}
candidate_skills = {"Python", "SQL", "AWS", "Snowflake"}
is_sub = required_skills.issubset(candidate_skills)
is_super = candidate_skills.issuperset(required_skills)
print(is_sub, is_super)
```

```python
# Q7
python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}
all_three = python & sql & aws
py_sql_not_aws = (python & sql) - aws
at_least_two = (python & sql) | (sql & aws) | (python & aws)
print(all_three, py_sql_not_aws, at_least_two)
```

```python
# Q8
day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}
all_days = day1 & day2 & day3
at_least_two_days = (day1 & day2) | (day2 & day3) | (day1 & day3)
all_ids = day1 | day2 | day3
exactly_one_day = all_ids - at_least_two_days
print(all_days, exactly_one_day, at_least_two_days)
```

```python
# Q9
required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}
missing = required_permissions - user_permissions
extra = user_permissions - required_permissions
full_auth = required_permissions.issubset(user_permissions)
print(missing, extra, full_auth)
```

```python
# Q10
website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}
all_platforms = website & mobile & store
at_least_two_plat = (website & mobile) | (mobile & store) | (website & store)
exactly_two = at_least_two_plat - all_platforms
all_customers = website | mobile | store
exactly_one_plat = all_customers - at_least_two_plat
print(exactly_one_plat, exactly_two, all_platforms)
```
