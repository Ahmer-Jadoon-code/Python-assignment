```python
# Q5
marks = {"Ali": 85, "Sara": 92, "Ahmed": 78, "Zain": 88}
highest_mark = max(marks.values())
lowest_mark = min(marks.values())
average_mark = sum(marks.values()) / len(marks)
swapped_marks = list(zip(marks.values(), marks.keys()))
top_student = max(swapped_marks)[1]
print(highest_mark, lowest_mark, average_mark, top_student)
```

```python
# Q6
employees_q6 = {
    "E101": {"name": "Ali", "salary": 80000},
    "E102": {"name": "Sara", "salary": 120000},
    "E103": {"name": "Ahmed", "salary": 95000}
}
sara_salary = employees_q6["E102"]["salary"]
employees_q6["E101"]["salary"] = employees_q6["E101"]["salary"] + (employees_q6["E101"]["salary"] * 0.15)
total_payroll = employees_q6["E101"]["salary"] + employees_q6["E102"]["salary"] + employees_q6["E103"]["salary"]
print(sara_salary, employees_q6["E101"]["salary"], total_payroll)
```

```python
# Q8
order = {
    "laptop": {"price": 120000, "quantity": 2},
    "mouse": {"price": 1500, "quantity": 5},
    "keyboard": {"price": 3500, "quantity": 3}
}
laptop_subtotal = order["laptop"]["price"] * order["laptop"]["quantity"]
mouse_subtotal = order["mouse"]["price"] * order["mouse"]["quantity"]
keyboard_subtotal = order["keyboard"]["price"] * order["keyboard"]["quantity"]
grand_total = laptop_subtotal + mouse_subtotal + keyboard_subtotal
total_after_discount = grand_total - (grand_total * 0.10)
print("Subtotals:", laptop_subtotal, mouse_subtotal, keyboard_subtotal)
print("Totals:", grand_total, total_after_discount)
```

```python
# Q9
employees_q9 = {
    "Ali": {"python": 90, "sql": 85, "aws": 88},
    "Sara": {"python": 95, "sql": 92, "aws": 90},
    "Ahmed": {"python": 80, "sql": 87, "aws": 84}
}
ali_avg = sum(employees_q9["Ali"].values()) / len(employees_q9["Ali"])
sara_avg = sum(employees_q9["Sara"].values()) / len(employees_q9["Sara"])
ahmed_avg = sum(employees_q9["Ahmed"].values()) / len(employees_q9["Ahmed"])

avg_scores = [(ali_avg, "Ali"), (sara_avg, "Sara"), (ahmed_avg, "Ahmed")]
top_performer = max(avg_scores)[1]

sql_scores = [
    (employees_q9["Ali"]["sql"], "Ali"),
    (employees_q9["Sara"]["sql"], "Sara"),
    (employees_q9["Ahmed"]["sql"], "Ahmed")
]
highest_sql = max(sql_scores)[0]
print("Averages:", ali_avg, sara_avg, ahmed_avg)
print("Top Performer & Highest SQL:", top_performer, highest_sql)
```

```python
# Q10
inventory = {
    "P101": {"name": "Laptop", "price": 120000, "stock": 5},
    "P102": {"name": "Mouse", "price": 1500, "stock": 20},
    "P103": {"name": "Keyboard", "price": 3500, "stock": 10}
}
laptop_val = inventory["P101"]["price"] * inventory["P101"]["stock"]
mouse_val = inventory["P102"]["price"] * inventory["P102"]["stock"]
keyboard_val = inventory["P103"]["price"] * inventory["P103"]["stock"]
total_inventory_value = laptop_val + mouse_val + keyboard_val

values_list = [
    (laptop_val, inventory["P101"]["name"]),
    (mouse_val, inventory["P102"]["name"]),
    (keyboard_val, inventory["P103"]["name"])
]
most_valuable = max(values_list)[1]

inventory["P101"]["stock"] = inventory["P101"]["stock"] - 2
inventory["P102"]["stock"] = inventory["P102"]["stock"] - 5

print("Value & Most Valuable:", total_inventory_value, most_valuable)
print("New Stock Laptop:", inventory["P101"]["stock"], "Mouse:", inventory["P102"]["stock"])
```
