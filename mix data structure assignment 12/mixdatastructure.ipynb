```python
# Q1
students = [
    {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
    {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
    {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]

ali_avg = sum(students[0]["marks"]) / len(students[0]["marks"])
sara_avg = sum(students[1]["marks"]) / len(students[1]["marks"])
ahmed_avg = sum(students[2]["marks"]) / len(students[2]["marks"])

top_student = max((ali_avg, students[0]["name"]), (sara_avg, students[1]["name"]), (ahmed_avg, students[2]["name"]))[1]
names = (students[0]["name"], students[1]["name"], students[2]["name"])
unique_marks = set(students[0]["marks"]) | set(students[1]["marks"]) | set(students[2]["marks"])

print(ali_avg, sara_avg, ahmed_avg)
print(top_student)
print(names)
print(unique_marks)
```

```python
# Q2
employees = {
    "Ali": {"Python", "SQL", "AWS"},
    "Sara": {"Python", "Databricks", "SQL"},
    "Ahmed": {"AWS", "Snowflake", "SQL"}
}

common_skills = employees["Ali"] & employees["Sara"] & employees["Ahmed"]
ali_only = employees["Ali"] - (employees["Sara"] | employees["Ahmed"])
all_skills = employees["Ali"] | employees["Sara"] | employees["Ahmed"]

req = {"Python", "SQL"}
qualified = []

if req.issubset(employees["Ali"]):
    qualified.append("Ali")
if req.issubset(employees["Sara"]):
    qualified.append("Sara")
if req.issubset(employees["Ahmed"]):
    qualified.append("Ahmed")

print(common_skills)
print(ali_only)
print(all_skills)
print(set(qualified))
```

```python
# Q3
sales = [
    ("Laptop", 120000, 2),
    ("Mouse", 1500, 5),
    ("Keyboard", 3500, 3)
]

rev1 = sales[0][1] * sales[0][2]
rev2 = sales[1][1] * sales[1][2]
rev3 = sales[2][1] * sales[2][2]
total_revenue = rev1 + rev2 + rev3

top_product = max((rev1, sales[0][0]), (rev2, sales[1][0]), (rev3, sales[2][0]))[1]
name_to_rev = {sales[0][0]: rev1, sales[1][0]: rev2, sales[2][0]: rev3}
prices = {sales[0][1], sales[1][1], sales[2][1]}

print(total_revenue)
print(top_product)
print(name_to_rev)
print(prices)
```

```python
# Q4
customers = {
    "C101": {"name": "Ali", "orders": [1200, 2500, 1800]},
    "C102": {"name": "Sara", "orders": [3000, 4500]},
    "C103": {"name": "Ahmed", "orders": [1500, 2000, 1000]}
}

c101_total = sum(customers["C101"]["orders"])
c102_total = sum(customers["C102"]["orders"])
c103_total = sum(customers["C103"]["orders"])

top_customer = max((c101_total, customers["C101"]["name"]), (c102_total, customers["C102"]["name"]), (c103_total, customers["C103"]["name"]))[1]
totals = (c101_total, c102_total, c103_total)

print(c101_total, c102_total, c103_total)
print(top_customer)
print(totals)
```

```python
# Q5
courses = {
    "Python": {"Ali", "Sara", "Ahmed"},
    "SQL": {"Sara", "Ahmed", "Zain"},
    "AWS": {"Ali", "Ahmed", "Zain"}
}

all_three = courses["Python"] & courses["SQL"] & courses["AWS"]
at_least_two = (courses["Python"] & courses["SQL"]) | (courses["SQL"] & courses["AWS"]) | (courses["Python"] & courses["AWS"])
exactly_two = at_least_two - all_three
all_students = courses["Python"] | courses["SQL"] | courses["AWS"]
exactly_one = all_students - at_least_two

print(exactly_two)
print(all_three)
print(exactly_one)
print(len(all_students))
```

```python
# Q6
transactions = [
    {"id": 101, "customer": "Ali", "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
    {"id": 102, "customer": "Sara", "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
    {"id": 103, "customer": "Ali", "items": [("Monitor", 45000, 1)]}
]

t1_rev = (transactions[0]["items"][0][1] * transactions[0]["items"][0][2]) + (transactions[0]["items"][1][1] * transactions[0]["items"][1][2])
t2_rev = (transactions[1]["items"][0][1] * transactions[1]["items"][0][2]) + (transactions[1]["items"][1][1] * transactions[1]["items"][1][2])
t3_rev = (transactions[2]["items"][0][1] * transactions[2]["items"][0][2])

total_rev = t1_rev + t2_rev + t3_rev
ali_spending = t1_rev + t3_rev
largest_trans = max(t1_rev, t2_rev, t3_rev)

unique_items = {
    transactions[0]["items"][0][0], transactions[0]["items"][1][0],
    transactions[1]["items"][0][0], transactions[1]["items"][1][0],
    transactions[2]["items"][0][0]
}

print(total_rev)
print(ali_spending)
print(largest_trans)
print(unique_items)
```

```python
# Q7
pipelines = {
    "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
    "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
    "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}

p1_total = sum(pipelines["pipeline_1"]["records"])
p2_total = sum(pipelines["pipeline_2"]["records"])
p3_total = sum(pipelines["pipeline_3"]["records"])

total_processed = p1_total + p2_total + p3_total
largest_pipe = max(p1_total, p2_total, p3_total)

failed = []
if pipelines["pipeline_1"]["status"] == "failed":
    failed.append("pipeline_1")
if pipelines["pipeline_2"]["status"] == "failed":
    failed.append("pipeline_2")
if pipelines["pipeline_3"]["status"] == "failed":
    failed.append("pipeline_3")

avg1 = p1_total / len(pipelines["pipeline_1"]["records"])
avg2 = p2_total / len(pipelines["pipeline_2"]["records"])
avg3 = p3_total / len(pipelines["pipeline_3"]["records"])

print(total_processed)
print(set(failed))
print(largest_pipe)
print(avg1, avg2, avg3)
```

```python
# Q8
accounts = {
    "A101": {"holder": "Ali", "transactions": [5000, 2000, 3000, -1000]},
    "A102": {"holder": "Sara", "transactions": [10000, 3000, -2000]},
    "A103": {"holder": "Ahmed", "transactions": [7000, 1500, 2500]}
}

bal1 = sum(accounts["A101"]["transactions"])
bal2 = sum(accounts["A102"]["transactions"])
bal3 = sum(accounts["A103"]["transactions"])

final_bals = {"A101": bal1, "A102": bal2, "A103": bal3}
highest = max(bal1, bal2, bal3)

w1 = min(accounts["A101"]["transactions"])
w2 = min(accounts["A102"]["transactions"])
w3 = min(accounts["A103"]["transactions"])

total_with = 0
if w1 < 0: total_with += w1
if w2 < 0: total_with += w2
if w3 < 0: total_with += w3

largest_dep = max(
    max(accounts["A101"]["transactions"]),
    max(accounts["A102"]["transactions"]),
    max(accounts["A103"]["transactions"])
)

print(final_bals)
print(highest)
print(total_with)
print(largest_dep)
```

```python
# Q9
users = {
    "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
    "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
    "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}

p1 = users["Ali"]["purchased"]
p2 = users["Sara"]["purchased"]
p3 = users["Ahmed"]["purchased"]

two_users = (p1 & p2) | (p2 & p3) | (p1 & p3)
ali_wish = users["Ali"]["wishlist"] - p1
shared_wish = users["Ali"]["wishlist"] & users["Sara"]["wishlist"] & users["Ahmed"]["wishlist"]
all_prod = p1 | p2 | p3 | users["Ali"]["wishlist"] | users["Sara"]["wishlist"] | users["Ahmed"]["wishlist"]

print(two_users)
print(ali_wish)
print(shared_wish)
print(all_prod)
```

```python
# Q10
company = {
    "departments": {
        "Engineering": {
            "employees": [
                ("Ali", 120000, {"Python", "AWS", "SQL"}),
                ("Sara", 150000, {"Python", "Databricks", "SQL"})
            ],
            "budget": 500000
        },
        "Analytics": {
            "employees": [
                ("Ahmed", 100000, {"SQL", "Power BI"}),
                ("Zain", 130000, {"Python", "SQL", "Tableau"})
            ],
            "budget": 400000
        }
    }
}

e1 = company["departments"]["Engineering"]["employees"][0]
e2 = company["departments"]["Engineering"]["employees"][1]
a1 = company["departments"]["Analytics"]["employees"][0]
a2 = company["departments"]["Analytics"]["employees"][1]

eng_pay = e1[1] + e2[1]
ana_pay = a1[1] + a2[1]
total_pay = eng_pay + ana_pay

top_emp = max((e1[1], e1[0]), (e2[1], e2[0]), (a1[1], a1[0]), (a2[1], a2[0]))[1]

shared_skills = e1[2] & e2[2] & a1[2] & a2[2]
eng_skills = e1[2] | e2[2]
ana_skills = a1[2] | a2[2]
eng_only = eng_skills - ana_skills

eng_budget = company["departments"]["Engineering"]["budget"] - eng_pay
ana_budget = company["departments"]["Analytics"]["budget"] - ana_pay

eng_avg = eng_pay / 2
ana_avg = ana_pay / 2
top_dept = max((eng_avg, "Engineering"), (ana_avg, "Analytics"))[1]

name_salary = {e1[0]: e1[1], e2[0]: e2[1], a1[0]: a1[1], a2[0]: a2[1]}
all_unique = eng_skills | ana_skills
budget_ok = (company["departments"]["Engineering"]["budget"] + company["departments"]["Analytics"]["budget"]) >= total_pay

print(eng_pay, ana_pay)
print(total_pay)
print(top_emp)
print(shared_skills)
print(eng_only)
print(eng_budget, ana_budget)
print(top_dept)
print(name_salary)
print(all_unique)
print(budget_ok)
```
